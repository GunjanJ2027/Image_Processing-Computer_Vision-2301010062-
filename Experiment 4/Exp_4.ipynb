```python
img =cv2.imread('/content/eagle_img.webp', 0)
```

```python
f = np.fft.fft2(img)
```

```python
fshift= np.fft.fftshift(f)
```

```python
row, cols=img.shape
```

```python
crow,ccol=row//2,cols//2
```

```python
mask_1p=np.zeros((row,cols), np.uint8)
```

```python
mask_1p[crow-30:crow+30, ccol-30:ccol+30]=1
```

```python
lp= np.fft.ifft2(np.fft.ifftshift(fshift*mask_1p))
```

```python
cv2.imwrite("ipf_eagle.png", np.abs(lp))
```

```python
mask_hp=1-mask_1p
```

```python
hp = np.fft.ifft2(np.fft.ifftshift(fshift*mask_hp))
```

```python
cv2.imwrite('hpf_eagle.png', np.abs(hp))
```
